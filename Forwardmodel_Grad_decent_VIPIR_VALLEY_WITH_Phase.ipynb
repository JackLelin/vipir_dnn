```python
valleytx = 0.037109375
valleyty = -0.037109375

v_fes = []
v_theta = []

npyfilename = 'JM91J_2016014141141.npy'
result_npy = np.load(valleyfolder + vipir_2_valley_dict[npyfilename], allow_pickle=True, fix_imports=True)[()]
valleyz = result_npy['hts']

for zz in valleyz/1000:
    r,lon,lat,dec,ha,aspect,B = jro0.aspect_txty(igrf_year,zz,valleytx,valleyty)
    v_fes += [(np.dot(B,B))**0.5*1e-9*1.602e-19/9.1093e-31]
    v_theta += [aspect]

valley_fes = np.array(v_fes,dtype=np.float64) /1e6/2/np.pi
valley_theta = np.array(v_theta) 

def phase_sim(valley_fps, valley_fes, valley_theta, valleyz):
    valleyNe = valley_fps**2/8.06 * 1e11*(valley_fps>0).astype(np.int_)
    Phase_sim = np.zeros_like(valley_fps)
    Phase_sim[1:] = 2.0*80.6/(3e8*5e7**2)*valley_fes[1:]*2*np.pi*1e6*np.abs(np.cos(valley_theta[1:]))*valleyNe[1:]*(valleyz[1:]-valleyz[:-1])*1e3
    Phase_sim = np.cumsum(Phase_sim)
    return Phase_sim

def phase_loss(Valley_phase, Phase_sim):
    offset2 = np.mean(Phase_sim[170:])
    phase_error = []
    for phase in Valley_phase:
        offset1 = np.mean(phase[170:])
        phase_error += [np.sum(np.abs(phase[170:] - offset1 - (Phase_sim[170:] - offset2))**2)]

    return np.min(phase_error)
```

```python
def dist_loss(input, model, radius = 20): # radius is the maxmum loss between two points. 
    loss = 0.0
    input1 = input/np.array([11,700]) * 100 # rescale both model and input to a 100 x 100 dimension
    model1 = model/np.array([11,700]) * 100 #
    fmin, fmax = np.min(model1[:,0]), np.max(model1[:,0])
    hmin, hmax = np.min(model1[:,1]), np.max(model1[:,1])
    for point in input1:
        if point[0] > 100 or point[1] > 100: # point is out of the bound
        # if point[0] < fmin or point[0] > fmax or point[1] < hmin or point[1] > hmax:
            continue
        error = np.min(np.sum((model1-point)**2, axis = 1))
        # error = np.clip(error**0.5,max= radius)
        loss += error
#     print(model1)
    return loss

def softplus(x,beta=7):
        return 1/beta*np.log(1+np.exp(x*beta))
def fps_parabolic(a,z):
    para1 = a[0]-((z-a[1])/a[2]/3)**2
    para2 = a[3]-((z-a[4])/a[5]/3)**2
    para3 = a[6]-((z-a[7])/a[8]/3)**2

    fps1 = (80.6*(softplus(para1) + softplus(para2) + softplus(para3)))**0.5
    # fps2 = (80.6*np.max(np.vstack([para1,para2,para3]), axis=0))**0.5
    return fps1#,fps2

def fps_spline(a,Z):
    a = np.cumsum(a)
    # spline = inter.PchipInterpolator(z,a)
    spline = inter.CubicSpline(z,a)
    fps = spline(Z)
    fps[fps < 0] = 0
    return fps

def evalutate_phase(x_vals, Valley_phase ,fps_func):
    valley_fps = fps_func(x_vals)
    p_sim = phase_sim(valley_fps, valley_fes, valley_theta, valleyz)
    p_loss = phase_loss(Valley_phase, p_sim)
    # print(p_loss)
    return p_loss

def evalutate(x_vals, targetO, targetX ,fps_func):
    fvsO,vhsO,fvsX,vhsX=OXmode_cython(fps_func(x_vals), fes, dz)
    o = np.vstack((fvsO,vhsO)).T
    x = np.vstack((fvsX,vhsX)).T
    loss  = dist_loss(targetO, o)
    loss += dist_loss(targetX, x)
    # print(loss)
    return loss

def differential(loss_fun, x_vals, dx : list = [0.05,0.1,0.2]):
    grad = 0
    df = np.zeros_like(x_vals)
    ei = np.zeros_like(x_vals)
    for ddx in dx:
        df *= 0
        ei *= 0
        for i in range(x_vals.shape[0]):
            ei[i] = ddx
            fp = loss_fun(x_vals+ei)
            fn = loss_fun(x_vals-ei)
            ei *= 0
            df[i] = (fp - fn) / (2 * ddx)
        grad += df / np.linalg.norm(df)
    return grad/len(dx), loss_fun(x_vals)
# def differential(loss_fun, x_vals, dx : list = [0.05,0.1,0.15]):
#     grad = 0
#     df = np.zeros_like(x_vals)
#     ei = np.zeros_like(x_vals)
#     for ddx in dx:
#         df *= 0
#         ei *= 0
#         for i in range(x_vals.shape[0]):
#             ei[i] = ddx * x_vals[i]
#             fp = loss_fun(x_vals+ei)
#             fn = loss_fun(x_vals-ei)
#             ei *= 0
#             df[i] = (fp - fn) / (2 * ddx)
#         grad += df / np.linalg.norm(df)
#     return grad/len(dx), loss_fun(x_vals)

def regularization_loss(x_vals):
    anchors = np.cumsum(x_vals)
    return np.sum(np.abs(anchors)**2)
```
