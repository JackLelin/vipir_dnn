```python
import glob

from matplotlib import pyplot as plt
import numpy as np
import scipy.interpolate as inter
import datetime
import scipy
from scipy import ndimage
from skimage import morphology

# import VIPIRread
# import VIPIRclean
from joblib import Parallel, delayed
import multiprocessing as mp
```

```python
from multiprocessing import Pool

outdir = 'test/'

orgdir = 'org/'
vipirfiles = glob.glob1(orgdir,'*.npy') 
vipirfiles.sort()

crop_region = np.zeros([512, 408])
crop_region[120:-10,30:390] = 1

def cleaning_process(idx):
    
    ionograms = np.load(orgdir+vipirfiles[idx], allow_pickle = True)[()]
    original = ionograms['original'] 

    original -= (np.min(original)-1)
    original[np.isnan(original)] = 1
    original = original / np.median(original)

    before = original
    temp1 = np.log10(before)

    sigma1 = 0.7
    sigma2 = 2.7

    gauss1 = ndimage.gaussian_filter(temp1,sigma1)
    gauss2 = ndimage.gaussian_filter(temp1,sigma2)
    out = gauss1-gauss2
    # out -= np.min(out)

    thresh = out > np.percentile(out,86,axis=0)
    
    temp3 = morphology.remove_small_objects(thresh, min_size=5, connectivity=1)

    mask_full = temp3 * crop_region

    fig=plt.figure(figsize=(10,7))      

    plt.subplot(2,1,1)
    im = plt.imshow(np.log10(np.abs(original)),cmap='binary',origin='lower',aspect='auto')
    im.axes.set_title(vipirfiles[idx]); im.axes.title.set_fontsize(10.0)
    plt.colorbar()

    plt.subplot(2,1,2)
    im = plt.imshow(mask_full,cmap='binary',origin='lower',aspect='auto')
    # im.axes.set_title('LT %.0f:%.0f:%.0f'%(vipirTimeSct['start_hour'][0]-5.0,vipirTimeSct['start_minute'][0],vipirTimeSct['start_second'][0]*1.0))
    im.axes.title.set_fontsize(10.0)
    plt.colorbar()

    # return
    print('saving figure', vipirfiles[idx])    
    fig.savefig(outdir+vipirfiles[idx][:-4],bbox_inches='tight')
    plt.close()

    np.save(outdir+vipirfiles[idx][:-4], {'original':ionograms['original'], 'mask_full':mask_full})

# cleaning_process(600)
# processes_pool = Pool(24)
# processes_pool.map(cleaning_process, range(40,len(vipirfiles)))
# processes_pool.close()

Parallel(n_jobs=mp.cpu_count()-4)(delayed(cleaning_process)(idx) for idx in range(0,len(vipirfiles)))
```
